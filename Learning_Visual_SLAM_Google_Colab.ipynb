````python
%%writefile dataset.h

// include/myslam/dataset.h
#ifndef MYSLAM_DATASET_H
#define MYSLAM_DATASET_H

#pragma once

#include "myslam/common_include.h"
#include "myslam/frame.h"

// Forward declaration of Frame to avoid heavy include here.
// It is assumed that Frame::Ptr is used in the next stage.
namespace myslam {
// struct Frame;
// typedef std::shared_ptr<Frame> FramePtr;

class Dataset {
public:
    EIGEN_MAKE_ALIGNED_OPERATOR_NEW;
    typedef std::shared_ptr<Dataset> Ptr;
    typedef Frame::Ptr FramePtr;   // ← This line is essential.

    Dataset(const std::string& dataset_path);

    // Reading dataset information (paths, calibration in the full version)
    bool Init();

    // Get the next frame; returns nullptr if finished.
    FramePtr NextFrame();

    // Number of images
    int ImageCount() const { return static_cast<int>(left_images_.size()); }

private:
    std::string dataset_path_;
    int current_image_index_ = 0;

    // Left/Right Image Path List
    std::vector<std::string> left_images_;
    std::vector<std::string> right_images_;
};

} // namespace myslam

#endif // MYSLAM_DATASET_H
````

### **main.cpp**

````python
%%writefile dataset.cpp

// src/dataset.cpp
#include "myslam/dataset.h"
#include "myslam/common_include.h"
#include "myslam/frame.h"   // Applying real frame

namespace myslam {

Dataset::Dataset(const std::string& dataset_path) : dataset_path_(dataset_path) {}

bool Dataset::Init() {
    namespace fs = std::filesystem;
    std::string left_dir = dataset_path_ + "/image_0";
    std::string right_dir = dataset_path_ + "/image_1";

    if (!fs::exists(left_dir) || !fs::exists(right_dir)) {
        std::cerr << "Dataset folders not found: " << left_dir << " or " << right_dir << std::endl;
        return false;
    }

    left_images_.clear();
    right_images_.clear();

    for (auto &p : fs::directory_iterator(left_dir)) left_images_.push_back(p.path().string());
    for (auto &p : fs::directory_iterator(right_dir)) right_images_.push_back(p.path().string());

    std::sort(left_images_.begin(), left_images_.end());
    std::sort(right_images_.begin(), right_images_.end());

    if (left_images_.size() != right_images_.size()) {
        std::cerr << "Left and right image counts differ!" << std::endl;
        size_t n = std::min(left_images_.size(), right_images_.size());
        left_images_.resize(n);
        right_images_.resize(n);
    }

    std::cout << "Dataset init: found " << left_images_.size() << " stereo pairs.\n";
    return true;
}

Dataset::FramePtr Dataset::NextFrame() {
    if (current_image_index_ >= static_cast<int>(left_images_.size())) return nullptr;

    cv::Mat left = cv::imread(left_images_[current_image_index_], cv::IMREAD_GRAYSCALE);
    cv::Mat right = cv::imread(right_images_[current_image_index_], cv::IMREAD_GRAYSCALE);

    if (left.empty() || right.empty()) {
        std::cerr << "Failed to read images at index " << current_image_index_ << std::endl;
        return nullptr;
    }

    Frame::Ptr f = Frame::CreateFrame();
    f->left_img = left;
    f->right_img = right;

    current_image_index_++;
    return f;
}

} // namespace myslam
````

````python
%%writefile main.cpp

// src/main.cpp
#include "myslam/common_include.h"
#include "myslam/dataset.h"
#include "myslam/frame.h"
#include <chrono>
#include <thread>

using namespace myslam;

int main(int argc, char** argv) {
    if (argc != 2) {
        std::cout << "Usage: run_dataset <dataset_directory>" << std::endl;
        return -1;
    }

    std::string dataset_path = argv[1];
    Dataset::Ptr ds(new Dataset(dataset_path));
    if (!ds->Init()) {
        std::cerr << "Dataset init failed!" << std::endl;
        return -1;
    }

    std::cout << "Total frames: " << ds->ImageCount() << std::endl;

    while (true) {
        auto frame = ds->NextFrame();
        if (!frame) break;

        // cv::imshow("left", frame->left_img);
        // cv::imshow("right", frame->right_img);

        cv::imwrite("left.png", frame->left_img);
        cv::imwrite("right.png", frame->right_img);

        int key = cv::waitKey(30);
        if (key == 27) break; // ESC
    }

    std::cout << "All frames processed." << std::endl;
    return 0;

}
````

# **Step 2: Feature Extraction**

**Current SLAM Pipeline**

**Dataset → Frame → Feature Extraction → Feature Matching → Pose Estimation → Mapping**

At this point, we are currently here:

```text
[✓] Load stereo frames
[✓] Display images
[ ] Frame structure (complete)
[ ] Feature extraction
[ ] Feature matching
```

Therefore, the next step is:

**Phase 1 — Complete the `Frame` Class**

The `Frame` class should be able to:

| Capability | Status |
|---|---|
| Store left and right images | ✓ |
| Store keypoints | In progress |
| Store descriptors | In progress |
| Store the frame ID | ✓ |
| Provide a constructor / frame creation mechanism | ✓ |

Once the `Frame` structure is complete, we can move on to **feature extraction** and then **feature matching**.

## **Headers**

````python
%cd /content/MySLAM_Ch13/include/myslam
````

````python
%%writefile frame.h

// include/myslam/frame.h
#pragma once
#include "common_include.h"

namespace myslam {

struct Frame {
    typedef std::shared_ptr<Frame> Ptr;
    long unsigned int id;
    cv::Mat left_img, right_img;

    // New fields
    std::vector<cv::KeyPoint> keypoints_left, keypoints_right;
    cv::Mat descriptors_left, descriptors_right;

    static Frame::Ptr CreateFrame();

    Frame() {}

    // Feature extraction
    void ExtractFeatures();
};

}  // namespace myslam
````

**Automatic Resource Management**

OpenCV automatically stores the appropriate values inside these variables.

There is no need to manually define a custom constructor or destructor for them because:

- `cv::Mat` and `std::vector` manage their own resources automatically.
- C++ automatically initializes and destroys these objects when their lifetime begins and ends.
- This behavior follows the **RAII (Resource Acquisition Is Initialization)** principle, which helps manage memory and resources safely.

## **cpp**

````python
%cd /content/MySLAM_Ch13/src
````

````python
%%writefile frame.cpp

#include "myslam/frame.h"

namespace myslam {

static long unsigned int global_frame_id = 0;

Frame::Ptr Frame::CreateFrame() {
    Frame::Ptr new_frame(new Frame());
    new_frame->id = global_frame_id++;
    return new_frame;
}

void Frame::ExtractFeatures() {
    cv::Ptr<cv::Feature2D> orb = cv::ORB::create(200); // تعداد featureها: بعداً تنظیم می‌کنیم

    // Extraction from the left image
    orb->detectAndCompute(left_img, cv::Mat(), keypoints_left, descriptors_left);

    // If we have stereo → we also extract from the right image.
    if (!right_img.empty()) {
        orb->detectAndCompute(right_img, cv::Mat(), keypoints_right, descriptors_right);
    }
}

}
````

# **Step 3: Feature Matching**

Feature Matching

## **Headers**

````python
%cd /content/MySLAM_Ch13/include/myslam
````

````python
%%writefile frame.h

// include/myslam/frame.h
#pragma once
#include "common_include.h"

namespace myslam {

struct Frame {
    typedef std::shared_ptr<Frame> Ptr;
    long unsigned int id;
    cv::Mat left_img, right_img;

    // New fields
    std::vector<cv::KeyPoint> keypoints_left, keypoints_right;
    cv::Mat descriptors_left, descriptors_right;

    static Frame::Ptr CreateFrame();

    Frame() {}

    // Feature extraction
    void ExtractFeatures();

    // Feature matcing
    void MatchFeaturesStereo(std::vector<cv::DMatch> &matches);

};

}  // namespace myslam
````

## **cpp**

````python
%cd /content/MySLAM_Ch13/src
````

````python
%%writefile frame.cpp

#include "myslam/frame.h"

namespace myslam {

static long unsigned int global_frame_id = 0;

Frame::Ptr Frame::CreateFrame() {
    Frame::Ptr new_frame(new Frame());
    new_frame->id = global_frame_id++;
    return new_frame;
}

void Frame::ExtractFeatures() {
    cv::Ptr<cv::Feature2D> orb = cv::ORB::create(200); // تعداد featureها: بعداً تنظیم می‌کنیم

    // استخراج از تصویر چپ
    orb->detectAndCompute(left_img, cv::Mat(), keypoints_left, descriptors_left);

    // اگر stereo داریم → از تصویر راست نیز استخراج می‌کنیم
    if (!right_img.empty()) {
        orb->detectAndCompute(right_img, cv::Mat(), keypoints_right, descriptors_right);
    }
}

void Frame::MatchFeaturesStereo(std::vector<cv::DMatch>& matches) {
    if (descriptors_left.empty() || descriptors_right.empty()) return;

    cv::BFMatcher matcher(cv::NORM_HAMMING, true); // crossCheck = true
    matcher.match(descriptors_left, descriptors_right, matches);

    // Sort matches based on distance (quality)
    std::sort(matches.begin(), matches.end(), [](const cv::DMatch& m1, const cv::DMatch& m2){
        return m1.distance < m2.distance;
    });
}


}
````

````python
%%writefile main.cpp

// src/main.cpp
#include "myslam/common_include.h"
#include "myslam/dataset.h"
#include "myslam/frame.h"
#include <chrono>
#include <thread>

using namespace myslam;

int main(int argc, char** argv) {
    if (argc != 2) {
        std::cout << "Usage: run_dataset <dataset_directory>" << std::endl;
        return -1;
    }

    std::string dataset_path = argv[1];
    Dataset::Ptr ds(new Dataset(dataset_path));
    if (!ds->Init()) {
        std::cerr << "Dataset init failed!" << std::endl;
        return -1;
    }

    std::cout << "Total frames: " << ds->ImageCount() << std::endl;

    while (true) {
        auto frame = ds->NextFrame();
        if (!frame) break;

        if (frame->left_img.empty() || frame->right_img.empty()) {
            std::cout << "Error reading images!" << std::endl;
            return -1;
        }

        // cv::imshow("left", frame->left_img);
        // cv::imshow("right", frame->right_img);

        if (frame->left_img.empty()) {
        std::cout << "Cannot load image!" << std::endl;
        return -1;
        }

        std::cout << "Frame ID: " << frame->id << std::endl;

        frame->ExtractFeatures();

        std::cout << "Extracted keypoints: " << frame->keypoints_left.size() << std::endl;
        std::cout << "Descriptor size: " << frame->descriptors_left.rows << " x "
                  << frame->descriptors_left.cols << std::endl;

        // Display as test
        cv::Mat vis;
        cv::drawKeypoints(frame->left_img, frame->keypoints_left, vis, cv::Scalar(0,255,0));
        // cv::imshow("Features", vis);
        cv::imwrite("Features.png", vis);
        cv::waitKey(0);


        // Feature Matching
        std::vector<cv::DMatch> matches;
        frame->MatchFeaturesStereo(matches);
        std::cout << "Found matches: " << matches.size() << std::endl;
        // Display for testing
        // Visualization
        cv::Mat viss;
        cv::drawMatches(frame->left_img, frame->keypoints_left,
                        frame->right_img, frame->keypoints_right,
                        matches, viss);
        //cv::imshow("Stereo Matches", viss);
        cv::imwrite("Stereo Matches.png", viss);
        cv::waitKey(0);


        cv::imwrite("left.png", frame->left_img);
        cv::imwrite("right.png", frame->right_img);
        //cv::imwrite("left_" + std::to_string(frame->id) + ".png", frame->left_img);
        //cv::imwrite("right_" + std::to_string(frame->id) + ".png", frame->left_img);


        int key = cv::waitKey(0);
        if (key == 27) break; // ESC
    }

    std::cout << "All frames processed." << std::endl;
    return 0;
}
````

# **Step 4: Improve Feature Quality (Add ROI and Mask)**

Before moving further into **depth estimation, triangulation, and pose estimation**, we should improve the quality of the extracted features.

The reason is simple: poor-quality or noisy features can later cause:

- poor depth estimation,
- unstable PnP,
- increased drift,
- more outliers and noisy MapPoints.

So before using features in the rest of the SLAM pipeline, it is important to make the feature extraction stage more selective and robust.

---

**1. Add ROI / Mask**

In KITTI images, some regions may contain many features but are not very useful for SLAM.

For example:

- sky → very little useful texture,
- lower parts of the image → parts of the vehicle or road,
- reflections or unstable image regions.

These features may add computational cost without improving tracking or mapping quality.

A simple approach is to define an **ROI (Region of Interest)** and run the feature detector only inside the useful part of the image.

Example:

```cpp
cv::Rect roi(
    0,
    0,
    left_img.cols,
    left_img.rows * 0.8
);

cv::Mat left_roi = left_img(roi);
```

Then the feature detector can be applied only to `left_roi`.

This helps keep features in more informative image regions.

---

**2. Sort KeyPoints by Response**

Not all detected keypoints have the same quality.

OpenCV provides:

```cpp
keypoint.response
```

which represents the strength or quality of a detected keypoint.

So we can sort the keypoints by response and keep only the strongest ones:

```cpp
std::sort(
    keypoints.begin(),
    keypoints.end(),
    [](const cv::KeyPoint& a, const cv::KeyPoint& b) {
        return a.response > b.response;
    }
);

if (keypoints.size() > num_features_)
    keypoints.resize(num_features_);
```

This keeps the highest-quality features and removes weaker ones.

As a result, we usually get:

- better feature matching,
- fewer outliers,
- more stable pose estimation,
- and more reliable MapPoints.

---

**3. Move Feature Extraction to the `Frontend` Class**

Currently, feature extraction is performed inside `Frame`.

However, from a design perspective, this is not ideal.

A `Frame` should mainly act as a **data container**, while the `Frontend` should perform processing tasks such as:

- feature detection,
- feature description,
- feature matching,
- tracking,
- pose estimation.

A cleaner design would therefore be:

```text
Dataset → Frontend → Backend
             |
           Frame
```

So instead of:

```cpp
frame->ExtractFeatures();
```

we can use:

```cpp
bool Frontend::DetectAndCompute(Frame::Ptr frame);
```

This keeps the responsibilities more clearly separated.

**Why Is This Better?**

- The `Frame` class becomes simpler.
- Feature detection settings can be changed more easily.
- Different detectors can be replaced or configured inside `Frontend`.
- Dependencies become cleaner.
- The code follows the **Single Responsibility Principle** more closely.

---

**4. Summary of Changes**

The planned changes are:

1. Add or update:

```text

```

2. Add:

```text

```

3. Remove or stop using:

```cpp
Frame::ExtractFeatures()
```

from `Frame`, if feature extraction is fully moved into `Frontend`.

If needed, it could temporarily remain as a wrapper, but removing it gives cleaner responsibility separation.

4. Update `main.cpp` to use `Frontend` for:

- feature detection,
- feature matching.

5. Update `CMakeLists.txt` and add:

```text

```

to the executable/source list.

---

**Where This Fits in the Pipeline**

After this step, the intended flow becomes:

**Stereo Matching → Disparity / Depth → Triangulation → MapPoints → Map**

and later:

**Feature Matching with Map / KeyFrames → PnP → New Camera Pose**

So improving feature quality here directly improves everything that comes after it.

**Parameter Tuning and Debugging Tips**

- **`roi_top_fraction_`**  
  Values between `0.0` and `0.2` are commonly reasonable.  
  For KITTI, around `0.10–0.15` is a good starting point because it removes some of the distant sky region.  
  For indoor datasets, setting it to `0.0` may be better.

- **`num_features_`**  
  A range of about `200–500` is a good starting point.  
  For better robustness, try around `500` features, but keep in mind that this increases computational cost.

- **Smarter Masks**  
  Later, the mask can be improved by detecting and suppressing uninformative regions, for example:
  - very bright regions near the top of the image,
  - sky-colored regions,
  - other visually unreliable areas.

- **Geometric Validation After Matching**  
  After feature matching, apply an **epipolar / Fundamental Matrix check** or **RANSAC** to reject incorrect matches and remove outliers.

  This step is especially important for obtaining reliable **disparity** and accurate **triangulation**.

````python
%%writefile frontend.h
// include/myslam/frontend.h
#pragma once
#include "common_include.h"
#include "myslam/frame.h"

namespace myslam {

class Frontend {
public:
    typedef std::shared_ptr<Frontend> Ptr;
    EIGEN_MAKE_ALIGNED_OPERATOR_NEW

    Frontend(int num_features = 200, double roi_top_fraction = 0.0);
    ~Frontend() = default;

    // Setting
    void SetNumFeatures(int n) { num_features_ = n; }
    void SetROITopFraction(double f) { roi_top_fraction_ = f; } // fraction [0..0.5] to crop top

    // Main API
    bool DetectAndCompute(Frame::Ptr frame); // Feature extraction with mask/ROI and sort by response
    void MatchStereo(Frame::Ptr frame, std::vector<cv::DMatch> &matches); // match Between Left and Right

private:
    int num_features_;
    double roi_top_fraction_; // What fraction of the top of the image should we remove?
    cv::Ptr<cv::ORB> orb_;
    cv::Ptr<cv::GFTTDetector> gftt_; // optional
    // helper
    cv::Rect ComputeROI(const cv::Mat &img) const;
    void KeepBestKeypoints(std::vector<cv::KeyPoint> &kps, cv::Mat &descriptors, int n);
};

} // namespace myslam
````

**Why Is `KeepBestKeypoints` Declared as `private`?**

Because this function is part of the class’s **internal logic**.

It:

- should only be used internally by the `Frontend`,
- is an implementation detail of feature processing,
- should not be called directly by other parts of the SLAM system.

If it were declared as `public`, other components could use it directly, which could lead to unnecessary coupling or architectural misuse.

In simple terms:

> `KeepBestKeypoints()` is not part of the public API of `Frontend`; it is an internal helper used during **feature preprocessing**.

### **frontend.cpp**

````python
%%writefile frontend.cpp

// src/frontend.cpp
#include "myslam/frontend.h"
#include <numeric>

namespace myslam {

Frontend::Frontend(int num_features, double roi_top_fraction)
    : num_features_(num_features), roi_top_fraction_(roi_top_fraction) {
    orb_ = cv::ORB::create(num_features_);
    gftt_ = cv::GFTTDetector::create(num_features_ * 2, 0.01, 20);
}

cv::Rect Frontend::ComputeROI(const cv::Mat &img) const {
    if (roi_top_fraction_ <= 0.0) return cv::Rect(0,0, img.cols, img.rows);
    int y = static_cast<int>(img.rows * roi_top_fraction_);
    int h = img.rows - y;
    return cv::Rect(0, y, img.cols, h);
}

void Frontend::KeepBestKeypoints(std::vector<cv::KeyPoint> &kps, cv::Mat &descriptors, int n) {
    if ((int)kps.size() <= n) return;

    // create indices sorted by response descending
    std::vector<int> idx(kps.size());
    std::iota(idx.begin(), idx.end(), 0);
    std::sort(idx.begin(), idx.end(), [&](int a, int b) {
        return kps[a].response > kps[b].response;
    });
    // pick top-n
    std::vector<cv::KeyPoint> kps2;
    cv::Mat desc2;
    kps2.reserve(n);
    for (int i = 0; i < n; ++i) {
        kps2.push_back(kps[idx[i]]);
        if (!descriptors.empty()) desc2.push_back(descriptors.row(idx[i]));
    }
    kps.swap(kps2);
    if (!desc2.empty()) descriptors = desc2;
}

bool Frontend::DetectAndCompute(Frame::Ptr frame) {
    if (!frame) return false;
    if (frame->left_img.empty()) return false;

    // ROI (crop top if configured)
    cv::Rect roi = ComputeROI(frame->left_img);
    cv::Mat left_roi = frame->left_img(roi);

    // Optional: create mask (exclude top small stripe or other rules)
    cv::Mat mask = cv::Mat::ones(left_roi.size(), CV_8U) * 255;
    // Example: exclude top 5% of roi (if you want more advanced mask, change here)
    // int exclude_h = static_cast<int>(0.05 * left_roi.rows);
    // if (exclude_h > 0) mask(cv::Rect(0,0,left_roi.cols, exclude_h)) = 0;

    // Detect + Compute on ROI
    std::vector<cv::KeyPoint> kps_left;
    cv::Mat desc_left;
    orb_->detectAndCompute(left_roi, mask, kps_left, desc_left);

    // adjust keypoint coordinates to original image coords
    for (auto &kp : kps_left) {
        kp.pt.x += roi.x;
        kp.pt.y += roi.y;
    }

    // Right image: detect on same ROI (or full image); here use same ROI for symmetry
    cv::Mat right_roi;
    std::vector<cv::KeyPoint> kps_right;
    cv::Mat desc_right;
    if (!frame->right_img.empty()) {
        right_roi = frame->right_img(roi);
        orb_->detectAndCompute(right_roi, mask, kps_right, desc_right);
        for (auto &kp : kps_right) {
            kp.pt.x += roi.x;
            kp.pt.y += roi.y;
        }
    }

    // Keep best by response
    KeepBestKeypoints(kps_left, desc_left, num_features_);
    if (!kps_right.empty()) KeepBestKeypoints(kps_right, desc_right, num_features_);

    // assign to frame
    frame->keypoints_left = std::move(kps_left);
    frame->descriptors_left = std::move(desc_left);
    frame->keypoints_right = std::move(kps_right);
    frame->descriptors_right = std::move(desc_right);

    return true;
}

void Frontend::MatchStereo(Frame::Ptr frame, std::vector<cv::DMatch> &matches) {
    matches.clear();
    if (!frame) return;
    if (frame->descriptors_left.empty() || frame->descriptors_right.empty()) return;

    cv::BFMatcher matcher(cv::NORM_HAMMING, true); // crossCheck = true
    matcher.match(frame->descriptors_left, frame->descriptors_right, matches);

    // sort by distance (best first)
    std::sort(matches.begin(), matches.end(), [](const cv::DMatch &a, const cv::DMatch &b){
        return a.distance < b.distance;
    });

    // optional: keep top ratio
    int keep = std::min<int>(matches.size(), num_features_);
    matches.resize(keep);
}

} // namespace myslam
````

**Why Do We Use `std::move`?**

`std::move` does not perform a normal copy.

It indicates that the resources owned by an object may be **transferred** to another object instead of being copied.

For example:

```cpp
frame->keypoints_left = std::move(kps_left);
```

This means:

- The data owned by `kps_left` is moved into:

```cpp
frame->keypoints_left
```

- This can avoid an unnecessary copy of the entire `keypoints` vector.
- For large containers, this is usually more efficient and is considered good modern C++ practice.

After the move operation, `kps_left` should generally no longer be relied on for its previous contents.

**Important C++ Detail**

A moved-from object is **not undefined**. It remains in a **valid but unspecified state**.

So after:

```cpp
std::move(kps_left)
```

`kps_left` may often become empty, but this is not something we should assume unless the type explicitly guarantees it.

> **Move semantics = transfer resources when possible, instead of performing an expensive copy.**

# **Step 5: Improve Feature Quality (Angle Filtering)**

**ROI Mask**

We have already applied an **ROI mask**.

Its purpose is to:

- restrict the region in which we search for keypoints,
- remove uninformative or useless image regions,
- and avoid extracting features from areas that are unlikely to produce reliable depth information.

---

**What Is Angle Filtering?**

This idea is related to geometric consistency in **stereo vision**.

In a rectified stereo camera setup, because the left and right cameras have a mainly horizontal baseline, the corresponding **epipolar lines are horizontal**.

Therefore, a feature detected in the left image should appear at approximately the same vertical image coordinate in the right image.

A stereo match may be considered suspicious or rejected if:

1. The vertical difference between the two matched keypoints is too large.
2. The feature orientations are strongly inconsistent.
3. The local gradient or ORB orientation is inconsistent with the expected stereo correspondence.

These checks can be used as part of **outlier filtering** before later stages such as:

**RANSAC → Pose Estimation**

**Important Note**

The most fundamental stereo constraint here is the **vertical alignment / epipolar constraint**.

Feature orientation or “angle filtering” is an additional consistency check; it is not itself the core epipolar geometry rule.

### **frontend.cpp**

````python
%%writefile frontend.cpp

// src/frontend.cpp
#include "myslam/frontend.h"
#include <numeric>

namespace myslam {

Frontend::Frontend(int num_features, double roi_top_fraction)
    : num_features_(num_features), roi_top_fraction_(roi_top_fraction) {
    orb_ = cv::ORB::create(num_features_);
    gftt_ = cv::GFTTDetector::create(num_features_ * 2, 0.01, 20);
}

cv::Rect Frontend::ComputeROI(const cv::Mat &img) const {
    if (roi_top_fraction_ <= 0.0) return cv::Rect(0,0, img.cols, img.rows);
    int y = static_cast<int>(img.rows * roi_top_fraction_);
    int h = img.rows - y;
    return cv::Rect(0, y, img.cols, h);
}

void Frontend::KeepBestKeypoints(std::vector<cv::KeyPoint> &kps, cv::Mat &descriptors, int n) {
    if ((int)kps.size() <= n) return;

    // create indices sorted by response descending
    std::vector<int> idx(kps.size());
    std::iota(idx.begin(), idx.end(), 0);
    std::sort(idx.begin(), idx.end(), [&](int a, int b) {
        return kps[a].response > kps[b].response;
    });
    // pick top-n
    std::vector<cv::KeyPoint> kps2;
    cv::Mat desc2;
    kps2.reserve(n);
    for (int i = 0; i < n; ++i) {
        kps2.push_back(kps[idx[i]]);
        if (!descriptors.empty()) desc2.push_back(descriptors.row(idx[i]));
    }
    kps.swap(kps2);
    if (!desc2.empty()) descriptors = desc2;
}

bool Frontend::DetectAndCompute(Frame::Ptr frame) {
    if (!frame) return false;
    if (frame->left_img.empty()) return false;

    // ROI (crop top if configured)
    cv::Rect roi = ComputeROI(frame->left_img);
    cv::Mat left_roi = frame->left_img(roi);

    // Optional: create mask (exclude top small stripe or other rules)
    cv::Mat mask = cv::Mat::ones(left_roi.size(), CV_8U) * 255;
    // Example: exclude top 5% of roi (if you want more advanced mask, change here)
    // int exclude_h = static_cast<int>(0.05 * left_roi.rows);
    // if (exclude_h > 0) mask(cv::Rect(0,0,left_roi.cols, exclude_h)) = 0;

    // Detect+Compute on ROI
    std::vector<cv::KeyPoint> kps_left;
    cv::Mat desc_left;
    orb_->detectAndCompute(left_roi, mask, kps_left, desc_left);

    // adjust keypoint coordinates to original image coords
    for (auto &kp : kps_left) {
        kp.pt.x += roi.x;
        kp.pt.y += roi.y;
    }

    // Right image: detect on same ROI (or full image); here use same ROI for symmetry
    cv::Mat right_roi;
    std::vector<cv::KeyPoint> kps_right;
    cv::Mat desc_right;
    if (!frame->right_img.empty()) {
        right_roi = frame->right_img(roi);
        orb_->detectAndCompute(right_roi, mask, kps_right, desc_right);
        for (auto &kp : kps_right) {
            kp.pt.x += roi.x;
            kp.pt.y += roi.y;
        }
    }

    // Keep best by response
    KeepBestKeypoints(kps_left, desc_left, num_features_);
    if (!kps_right.empty()) KeepBestKeypoints(kps_right, desc_right, num_features_);

    // assign to frame
    frame->keypoints_left = std::move(kps_left);
    frame->descriptors_left = std::move(desc_left);
    frame->keypoints_right = std::move(kps_right);
    frame->descriptors_right = std::move(desc_right);

    return true;
}

void Frontend::MatchStereo(Frame::Ptr frame, std::vector<cv::DMatch> &matches) {
    matches.clear();
    if (!frame) return;
    if (frame->descriptors_left.empty() || frame->descriptors_right.empty()) return;

    // Step 1: brute-force matching
    cv::BFMatcher matcher(cv::NORM_HAMMING, true); // crossCheck = true
    std::vector<cv::DMatch> raw_matches;
    matcher.match(frame->descriptors_left, frame->descriptors_right, raw_matches);

    // Step 2: filtering based on epipolar constraint and angle similarity
    const double max_y_diff = 2.0;
    const double max_angle_diff = 30.0;

    for (auto& m : raw_matches) {
        const auto& kpL = frame->keypoints_left[m.queryIdx];
        const auto& kpR = frame->keypoints_right[m.trainIdx];

        if (std::fabs(kpL.pt.y - kpR.pt.y) > max_y_diff)
            continue;

        if (std::fabs(kpL.angle - kpR.angle) > max_angle_diff)
            continue;

        matches.push_back(m);
    }

    // Step 3: sort by descriptor distance
    std::sort(matches.begin(), matches.end(), [](const cv::DMatch &a, const cv::DMatch &b){
        return a.distance < b.distance;
    });

    // Step 4: keep only N best if needed
    int keep = std::min<int>(matches.size(), num_features_);
    matches.resize(keep);

    std::cout << "[Stereo Match] Raw: " << raw_matches.size()
              << "  After Filter: " << matches.size() << std::endl;
}

} // namespace myslam
````

# **Step 6: Disparity and Triangulation**

**Next Step: Disparity and Triangulation**

At this point, we already have:

- Feature Detection
- Stereo Matching
- ROI Filtering
- Angle / Geometric Filtering

So the detected features and stereo correspondences are now much cleaner and more reliable.

The next step is:

**Disparity → Depth → Triangulation**

---

**Important Point Before Starting**

We currently have:

```cpp
std::vector<cv::DMatch> matches;
```

For each stereo match, we can access the corresponding keypoints in the left and right images:

```cpp
auto &kpL = frame->keypoints_left[m.queryIdx];
auto &kpR = frame->keypoints_right[m.trainIdx];
```

Then we compute the stereo disparity as:

```cpp
double disparity = kpL.pt.x - kpR.pt.x;
```

Because the stereo images are **rectified**, corresponding points are expected to have approximately the same vertical coordinate:

```text
yL ≈ yR
```

So disparity is mainly measured along the horizontal image axis.

**Important Validity Check**

If the disparity is too small, the estimated depth becomes very large and unreliable.

Therefore, very small or invalid disparity values should be rejected before depth estimation and triangulation.

So the flow becomes:

**Stereo Match → Disparity Check → Depth Estimation → 3D Point / MapPoint**

### **frontend.cpp**

````python
%%writefile frontend.cpp

// src/frontend.cpp
#include "myslam/frontend.h"
#include <numeric>

namespace myslam {

Frontend::Frontend(int num_features, double roi_top_fraction)
    : num_features_(num_features), roi_top_fraction_(roi_top_fraction) {
    orb_ = cv::ORB::create(num_features_);
    gftt_ = cv::GFTTDetector::create(num_features_ * 2, 0.01, 20);
}

cv::Rect Frontend::ComputeROI(const cv::Mat &img) const {
    if (roi_top_fraction_ <= 0.0) return cv::Rect(0,0, img.cols, img.rows);
    int y = static_cast<int>(img.rows * roi_top_fraction_);
    int h = img.rows - y;
    return cv::Rect(0, y, img.cols, h);
}

void Frontend::KeepBestKeypoints(std::vector<cv::KeyPoint> &kps, cv::Mat &descriptors, int n) {
    if ((int)kps.size() <= n) return;

    // create indices sorted by response descending
    std::vector<int> idx(kps.size());
    std::iota(idx.begin(), idx.end(), 0);
    std::sort(idx.begin(), idx.end(), [&](int a, int b) {
        return kps[a].response > kps[b].response;
    });
    // pick top-n
    std::vector<cv::KeyPoint> kps2;
    cv::Mat desc2;
    kps2.reserve(n);
    for (int i = 0; i < n; ++i) {
        kps2.push_back(kps[idx[i]]);
        if (!descriptors.empty()) desc2.push_back(descriptors.row(idx[i]));
    }
    kps.swap(kps2);
    if (!desc2.empty()) descriptors = desc2;
}

bool Frontend::DetectAndCompute(Frame::Ptr frame) {
    if (!frame) return false;
    if (frame->left_img.empty()) return false;

    // ROI (crop top if configured)
    cv::Rect roi = ComputeROI(frame->left_img);
    cv::Mat left_roi = frame->left_img(roi);

    // Optional: create mask (exclude top small stripe or other rules)
    cv::Mat mask = cv::Mat::ones(left_roi.size(), CV_8U) * 255;
    // Example: exclude top 5% of roi (if you want more advanced mask, change here)
    // int exclude_h = static_cast<int>(0.05 * left_roi.rows);
    // if (exclude_h > 0) mask(cv::Rect(0,0,left_roi.cols, exclude_h)) = 0;

    // Detect+Compute on ROI
    std::vector<cv::KeyPoint> kps_left;
    cv::Mat desc_left;
    orb_->detectAndCompute(left_roi, mask, kps_left, desc_left);

    // adjust keypoint coordinates to original image coords
    for (auto &kp : kps_left) {
        kp.pt.x += roi.x;
        kp.pt.y += roi.y;
    }

    // Right image: detect on same ROI (or full image); here use same ROI for symmetry
    cv::Mat right_roi;
    std::vector<cv::KeyPoint> kps_right;
    cv::Mat desc_right;
    if (!frame->right_img.empty()) {
        right_roi = frame->right_img(roi);
        orb_->detectAndCompute(right_roi, mask, kps_right, desc_right);
        for (auto &kp : kps_right) {
            kp.pt.x += roi.x;
            kp.pt.y += roi.y;
        }
    }

    // Keep best by response
    KeepBestKeypoints(kps_left, desc_left, num_features_);
    if (!kps_right.empty()) KeepBestKeypoints(kps_right, desc_right, num_features_);

    // assign to frame
    frame->keypoints_left = std::move(kps_left);
    frame->descriptors_left = std::move(desc_left);
    frame->keypoints_right = std::move(kps_right);
    frame->descriptors_right = std::move(desc_right);

    return true;
}


void Frontend::ComputeStereoDepth(Frame::Ptr frame, const std::vector<cv::DMatch> &matches) {

    const double fx = camera_left_->fx();
    const double fy = camera_left_->fy();
    const double cx = camera_left_->cx();
    const double cy = camera_left_->cy();
    const double baseline = camera_right_->baseline();

    frame->points_3d.clear();
    frame->points_3d.reserve(matches.size());

    for(const auto &m : matches) {
        const cv::KeyPoint &kpL = frame->keypoints_left[m.queryIdx];
        const cv::KeyPoint &kpR = frame->keypoints_right[m.trainIdx];

        double disparity = kpL.pt.x - kpR.pt.x;
        if(disparity <= 0.5) continue;

        double Z = fx * baseline / disparity;
        double X = (kpL.pt.x - cx) * Z / fx;
        double Y = (kpL.pt.y - cy) * Z / fy;

        frame->points_3d.emplace_back(Eigen::Vector3d(X, Y, Z));
    }

    std::cout << "[Depth] Valid 3D points: " << frame->points_3d.size() << std::endl;
}

void Frontend::SetCameras(Camera::Ptr cam_left, Camera::Ptr cam_right) {
    camera_left_ = cam_left;
    camera_right_ = cam_right;
}


void Frontend::MatchStereo(Frame::Ptr frame, std::vector<cv::DMatch> &matches) {
    matches.clear();
    if (!frame) return;
    if (frame->descriptors_left.empty() || frame->descriptors_right.empty()) return;

    // Step 1: brute-force matching
    cv::BFMatcher matcher(cv::NORM_HAMMING, true); // crossCheck = true
    std::vector<cv::DMatch> raw_matches;
    matcher.match(frame->descriptors_left, frame->descriptors_right, raw_matches);

    // Step 2: filtering based on epipolar constraint and angle similarity
    const double max_y_diff = 2.0;
    const double max_angle_diff = 30.0;

    for (auto& m : raw_matches) {
        const auto& kpL = frame->keypoints_left[m.queryIdx];
        const auto& kpR = frame->keypoints_right[m.trainIdx];

        if (std::fabs(kpL.pt.y - kpR.pt.y) > max_y_diff)
            continue;

        if (std::fabs(kpL.angle - kpR.angle) > max_angle_diff)
            continue;

        matches.push_back(m);
    }

    // Step 3: sort by descriptor distance
    std::sort(matches.begin(), matches.end(), [](const cv::DMatch &a, const cv::DMatch &b){
        return a.distance < b.distance;
    });

    // Step 4: keep only N best if needed
    int keep = std::min<int>(matches.size(), num_features_);
    matches.resize(keep);

    std::cout << "[Stereo Match] Raw: " << raw_matches.size()
              << "  After Filter: " << matches.size() << std::endl;

    // Step 5
    ComputeStereoDepth(frame, matches);

}

} // namespace myslam
````

### **main.cpp**

````python
%%writefile main.cpp

#include "myslam/common_include.h"
#include "myslam/dataset.h"
#include "myslam/frame.h"
#include "myslam/frontend.h"
#include "myslam/logger.h"

using namespace myslam;

int main(int argc, char** argv) {
    if (argc != 2) {
        Log(ERRORL, "Usage: run_dataset <dataset_directory>");
        return -1;
    }
    Dataset::Ptr ds(new Dataset(argv[1]));
    if (!ds->Init()) {
        Log(ERRORL, "Dataset init failed!");
        return -1;
    }
    Frontend::Ptr frontend(new Frontend(200, 0.12)); // keep 200 features, crop top 12%
    Log(INFO, "Total frames: " + std::to_string(ds->ImageCount()));

    auto camL = std::make_shared<myslam::Camera>(718.856, 718.856, 607.1928, 185.2157, 0.573);
    auto camR = std::make_shared<myslam::Camera>(718.856, 718.856, 607.1928, 185.2157, 0.573);

    frontend->SetCameras(camL, camR);

    int frame_id = 0;
    //std::cout << "frame_id: " << frame_id;
    int ii = 0;

    while (ii < 20) {
      auto frame = ds->NextFrame();
      if (!frame) {
      Log(ERRORL, "Frame is null!");
      break;
      }

      if (frame->left_img.empty() || frame->right_img.empty()) {
          Log(ERRORL, "Empty image at frame " + std::to_string(frame->id));
          return -1;
      }

      Log(INFO, "Images loaded, size L: " + std::to_string(frame->left_img.cols) + "x" + std::to_string(frame->left_img.rows) +
                " R: " + std::to_string(frame->right_img.cols) + "x" + std::to_string(frame->right_img.rows));


      ii = frame->id;
      Log(INFO, "Processing Frame: " + std::to_string(frame->id));

      // Check image validity
      if (frame->left_img.empty() || frame->right_img.empty()) {
      Log(ERRORL, "Empty image at frame " + std::to_string(frame->id));
      continue;
      }

      // Feature detection
      if (!frontend->DetectAndCompute(frame)) {
          Log(WARNING, "Detect failed for frame " + std::to_string(frame->id));
          continue;
      }

      // Check keypoints size
      if (frame->keypoints_left.empty() || frame->keypoints_right.empty()) {
          Log(WARNING, "No keypoints detected at frame " + std::to_string(frame->id));
          continue;
      }

      Log(DEBUGL, "Keypoints L: " + std::to_string(frame->keypoints_left.size())
                  + " R: " + std::to_string(frame->keypoints_right.size()));

      // Visualization
      cv::Mat vis;
      cv::drawKeypoints(frame->left_img, frame->keypoints_left, vis);
      cv::imwrite("debug/keypoints_left_" + std::to_string(frame->id) + ".png", vis);

      Log(INFO, "kps L = " + std::to_string(frame->keypoints_left.size()) +
              " | R = " + std::to_string(frame->keypoints_right.size()));


      std::vector<cv::DMatch> matches;
      frontend->MatchStereo(frame, matches);

      Log(DEBUGL, "Matches: " + std::to_string(matches.size()));

      if (matches.size() < 20) {
          Log(WARNING, "Too few stereo matches for frame " + std::to_string(frame->id));
          continue;
    }

    // Continue with triangulation later...
}

    Log(INFO, "All frames processed.");
    return 0;
}
````

````python
%%writefile mappoint.h

#pragma once
#include "common_include.h"

namespace myslam {

class MapPoint {
public:
    EIGEN_MAKE_ALIGNED_OPERATOR_NEW;
    using Ptr = std::shared_ptr<MapPoint>;

    unsigned long id_ = 0;          // Unique point ID
    Eigen::Vector3d pos_;           // 3D position (world coordinate)
    int observed_times_ = 0;        // Number of times observed in frames

    MapPoint() = default;

    MapPoint(long id, const Eigen::Vector3d& pos)
        : id_(id), pos_(pos) {}

    Eigen::Vector3d Pos() const { return pos_; }
    void SetPos(const Eigen::Vector3d& pos) { pos_ = pos; }

    void IncreaseObservation() { observed_times_++; }
};

} // namespace myslam

````

**What Is `observed_times_` Used For?**

`observed_times_` indicates how many times a **MapPoint** has been observed by KeyFrames or other frames.

Its value helps estimate how reliable and stable a landmark is.

- **Low value** — for example, observed only once: the point has a higher chance of being an outlier or an unreliable landmark.
- **High value**: the point has been observed repeatedly and is usually considered more stable and reliable.

**Uses in SLAM**

`observed_times_` can be useful for:

- selecting reliable points for **Bundle Adjustment (BA)**,
- deciding whether a MapPoint should be removed,
- assigning confidence or weight during optimization,
- identifying trustworthy landmarks for tracking and mapping.

For example:

```cpp
if (point->observed_times_ < 2)
    remove it;
```

This is a simple **MapPoint culling rule**: landmarks with too few observations may be removed from the map.

---



**Why Do We Have Two Constructors?**

```cpp
MapPoint() = default;

MapPoint(long id, const Eigen::Vector3d& pos)
    : id_(id), pos_(pos) {}
```

They allow us to create a `MapPoint` in two different ways.

**Default Constructor**

```cpp
MapPoint() = default;
```

Creates a `MapPoint` without explicitly providing an ID or position at construction time.

This is useful when the object must be created first and its data will be assigned later.

**Parameterized Constructor**

```cpp
MapPoint(long id, const Eigen::Vector3d& pos)
    : id_(id), pos_(pos) {}
```

Creates and initializes the `MapPoint` immediately with:

- an `id`
- a 3D position `pos`

For example:

```cpp
MapPoint mp(10, position);
```

**Why Is This Flexibility Useful?**

Sometimes all the required information is already available when the object is created, so the parameterized constructor is convenient.

In other cases, an object may need to be created first and filled later, for example in certain **factory or staged initialization patterns**.

> **Important C++ note:** `= default` does not automatically assign meaningful values to built-in members such as integers unless they already have in-class initializers. So members like `id_` should ideally have a safe default value if the default constructor is used.

### **frontend.cpp**

````python
%%writefile frontend.cpp

// src/frontend.cpp
#include "myslam/frontend.h"
#include <numeric>

namespace myslam {

Frontend::Frontend(int num_features, double roi_top_fraction)
    : num_features_(num_features), roi_top_fraction_(roi_top_fraction) {
    orb_ = cv::ORB::create(num_features_);
    gftt_ = cv::GFTTDetector::create(num_features_ * 2, 0.01, 20);
}

cv::Rect Frontend::ComputeROI(const cv::Mat &img) const {
    if (roi_top_fraction_ <= 0.0) return cv::Rect(0,0, img.cols, img.rows);
    int y = static_cast<int>(img.rows * roi_top_fraction_);
    int h = img.rows - y;
    return cv::Rect(0, y, img.cols, h);
}

void Frontend::KeepBestKeypoints(std::vector<cv::KeyPoint> &kps, cv::Mat &descriptors, int n) {
    if ((int)kps.size() <= n) return;
    // create indices sorted by response descending
    std::vector<int> idx(kps.size());
    std::iota(idx.begin(), idx.end(), 0);
    std::sort(idx.begin(), idx.end(), [&](int a, int b) {
        return kps[a].response > kps[b].response;
    });
    // pick top-n
    std::vector<cv::KeyPoint> kps2;
    cv::Mat desc2;
    kps2.reserve(n);
    for (int i = 0; i < n; ++i) {
        kps2.push_back(kps[idx[i]]);
        if (!descriptors.empty()) desc2.push_back(descriptors.row(idx[i]));
    }
    kps.swap(kps2);
    if (!desc2.empty()) descriptors = desc2;
}

bool Frontend::DetectAndCompute(Frame::Ptr frame) {
    if (!frame) return false;
    if (frame->left_img.empty()) return false;

    // ROI (crop top if configured)
    cv::Rect roi = ComputeROI(frame->left_img);
    cv::Mat left_roi = frame->left_img(roi);

    // Optional: create mask (exclude top small stripe or other rules)
    cv::Mat mask = cv::Mat::ones(left_roi.size(), CV_8U) * 255;
    // Example: exclude top 5% of roi (if you want more advanced mask, change here)
    // int exclude_h = static_cast<int>(0.05 * left_roi.rows);
    // if (exclude_h > 0) mask(cv::Rect(0,0,left_roi.cols, exclude_h)) = 0;

    // Detect+Compute on ROI
    std::vector<cv::KeyPoint> kps_left;
    cv::Mat desc_left;
    orb_->detectAndCompute(left_roi, mask, kps_left, desc_left);

    // adjust keypoint coordinates to original image coords
    for (auto &kp : kps_left) {
        kp.pt.x += roi.x;
        kp.pt.y += roi.y;
    }

    // Right image: detect on same ROI (or full image); here use same ROI for symmetry
    cv::Mat right_roi;
    std::vector<cv::KeyPoint> kps_right;
    cv::Mat desc_right;
    if (!frame->right_img.empty()) {
        right_roi = frame->right_img(roi);
        orb_->detectAndCompute(right_roi, mask, kps_right, desc_right);
        for (auto &kp : kps_right) {
            kp.pt.x += roi.x;
            kp.pt.y += roi.y;
        }
    }

    // Keep best by response
    KeepBestKeypoints(kps_left, desc_left, num_features_);
    if (!kps_right.empty()) KeepBestKeypoints(kps_right, desc_right, num_features_);

    // assign to frame
    frame->keypoints_left = std::move(kps_left);
    frame->descriptors_left = std::move(desc_left);
    frame->keypoints_right = std::move(kps_right);
    frame->descriptors_right = std::move(desc_right);

    return true;
}


void Frontend::ComputeStereoDepth(Frame::Ptr frame, const std::vector<cv::DMatch> &matches) {

    const double fx = camera_left_->fx();
    const double fy = camera_left_->fy();
    const double cx = camera_left_->cx();
    const double cy = camera_left_->cy();
    const double baseline = camera_right_->baseline();

    frame->points_3d.clear();
    frame->points_3d.reserve(matches.size());

    for(const auto &m : matches) {
        const cv::KeyPoint &kpL = frame->keypoints_left[m.queryIdx];
        const cv::KeyPoint &kpR = frame->keypoints_right[m.trainIdx];

        double disparity = kpL.pt.x - kpR.pt.x;
        if(disparity <= 0.5) continue;

        double Z = fx * baseline / disparity;
        double X = (kpL.pt.x - cx) * Z / fx;
        double Y = (kpL.pt.y - cy) * Z / fy;

        frame->points_3d.emplace_back(Eigen::Vector3d(X, Y, Z));
    }

    std::cout << "[Depth] Valid 3D points: " << frame->points_3d.size() << std::endl;
}

void Frontend::SetCameras(Camera::Ptr cam_left, Camera::Ptr cam_right) {
    camera_left_ = cam_left;
    camera_right_ = cam_right;
}


void Frontend::MatchStereo(Frame::Ptr frame, std::vector<cv::DMatch> &matches) {
    matches.clear();
    if (!frame) return;
    if (frame->descriptors_left.empty() || frame->descriptors_right.empty()) return;

    // Step 1: brute-force matching
    cv::BFMatcher matcher(cv::NORM_HAMMING, true); // crossCheck = true
    std::vector<cv::DMatch> raw_matches;
    matcher.match(frame->descriptors_left, frame->descriptors_right, raw_matches);

    // Step 2: filtering based on epipolar constraint and angle similarity
    const double max_y_diff = 2.0;
    const double max_angle_diff = 30.0;

    for (auto& m : raw_matches) {
        const auto& kpL = frame->keypoints_left[m.queryIdx];
        const auto& kpR = frame->keypoints_right[m.trainIdx];

        if (std::fabs(kpL.pt.y - kpR.pt.y) > max_y_diff)
            continue;

        if (std::fabs(kpL.angle - kpR.angle) > max_angle_diff)
            continue;

        matches.push_back(m);
    }

    // Step 3: sort by descriptor distance
    std::sort(matches.begin(), matches.end(), [](const cv::DMatch &a, const cv::DMatch &b){
        return a.distance < b.distance;
    });

    // Step 4: keep only N best if needed
    int keep = std::min<int>(matches.size(), num_features_);
    matches.resize(keep);

    std::cout << "[Stereo Match] Raw: " << raw_matches.size()
              << "  After Filter: " << matches.size() << std::endl;

   // Step 5
    ComputeStereoDepth(frame, matches);

    // Step 6 === Create MapPoints and insert into Map ===
    for (size_t i = 0; i < frame->points_3d.size(); ++i) {

        static unsigned long next_landmark_id_ = 0;
        Eigen::Vector3d Pw = frame->points_3d[i];

        // Depth sanity check
        if (Pw[2] < 0.3 || Pw[2] > 80.0) continue;

        auto mp = std::make_shared<MapPoint>(next_landmark_id_++, Pw);
        mp->IncreaseObservation();
        frame->AddLandmark(mp);
        map_->InsertMapPoint(mp);
    }

    // Mark this frame as keyframe (simple version)
    map_->InsertKeyFrame(frame);

    std::cout << "[MAP] KeyFrames: " << map_->KeyframeCount()
              << "  Landmarks: " << map_->MapPointsCount() << std::endl;

}

} // namespace myslam
````

````python
%%writefile main.cpp

#include "myslam/common_include.h"
#include "myslam/dataset.h"
#include "myslam/frame.h"
#include "myslam/frontend.h"
#include "myslam/logger.h"

using namespace myslam;

int main(int argc, char** argv) {
    if (argc != 2) {
        Log(ERRORL, "Usage: run_dataset <dataset_directory>");
        return -1;
    }
    Dataset::Ptr ds(new Dataset(argv[1]));
    if (!ds->Init()) {
        Log(ERRORL, "Dataset init failed!");
        return -1;
    }
    Frontend::Ptr frontend(new Frontend(200, 0.12)); // keep 200 features, crop top 12%
    Log(INFO, "Total frames: " + std::to_string(ds->ImageCount()));

    auto camL = std::make_shared<myslam::Camera>(718.856, 718.856, 607.1928, 185.2157, 0.573);
    auto camR = std::make_shared<myslam::Camera>(718.856, 718.856, 607.1928, 185.2157, 0.573);

    frontend->SetCameras(camL, camR);

    Map::Ptr map(new Map());
    frontend->SetMap(map);

    int frame_id = 0;
    //std::cout << "frame_id: " << frame_id;
    int ii = 0;

    while (ii < 20) {
      auto frame = ds->NextFrame();
      if (!frame) {
      Log(ERRORL, "Frame is null!");
      break;
      }

      if (frame->left_img.empty() || frame->right_img.empty()) {
          Log(ERRORL, "Empty image at frame " + std::to_string(frame->id));
          return -1;
      }

      Log(INFO, "Images loaded, size L: " + std::to_string(frame->left_img.cols) + "x" + std::to_string(frame->left_img.rows) +
                " R: " + std::to_string(frame->right_img.cols) + "x" + std::to_string(frame->right_img.rows));


      ii = frame->id;
      Log(INFO, "Processing Frame: " + std::to_string(frame->id));

      // Check image validity
      if (frame->left_img.empty() || frame->right_img.empty()) {
      Log(ERRORL, "Empty image at frame " + std::to_string(frame->id));
      continue;
      }


      // Feature detection
      if (!frontend->DetectAndCompute(frame)) {
          Log(WARNING, "Detect failed for frame " + std::to_string(frame->id));
          continue;
      }

      // Check keypoints size
      if (frame->keypoints_left.empty() || frame->keypoints_right.empty()) {
          Log(WARNING, "No keypoints detected at frame " + std::to_string(frame->id));
          continue;
      }

      Log(DEBUGL, "Keypoints L: " + std::to_string(frame->keypoints_left.size())
                  + " R: " + std::to_string(frame->keypoints_right.size()));

      // Visualization
      cv::Mat vis;
      cv::drawKeypoints(frame->left_img, frame->keypoints_left, vis);
      cv::imwrite("debug/keypoints_left_" + std::to_string(frame->id) + ".png", vis);

      Log(INFO, "kps L = " + std::to_string(frame->keypoints_left.size()) +
              " | R = " + std::to_string(frame->keypoints_right.size()));


      std::vector<cv::DMatch> matches;
      frontend->MatchStereo(frame, matches);

      Log(DEBUGL, "Matches: " + std::to_string(matches.size()));

      if (matches.size() < 10) {
          Log(WARNING, "Too few stereo matches for frame " + std::to_string(frame->id));
          continue;
      }

    // Continue with triangulation later...
  }

    Log(INFO, "All frames processed.");
    return 0;
}
````

````python
%%writefile frontend.h
// include/myslam/frontend.h
#pragma once
#include "common_include.h"
#include "frame.h"
#include "camera.h"
#include "map.h"

namespace myslam {

class Frontend {
public:
    typedef std::shared_ptr<Frontend> Ptr;

    void SetCameras(Camera::Ptr cam_left, Camera::Ptr cam_right);

    Frontend(int num_features = 200, double roi_top_fraction = 0.0);
    ~Frontend() = default;

    // تنظیمات
    void SetNumFeatures(int n) { num_features_ = n; }
    void SetROITopFraction(double f) { roi_top_fraction_ = f; } // fraction [0..0.5] to crop top

    // API اصلی
    bool DetectAndCompute(Frame::Ptr frame); // استخراج فیچر با mask/ROI و sort by response
    void MatchStereo(Frame::Ptr frame, std::vector<cv::DMatch> &matches); // match بین چپ و راست
    void ComputeStereoDepth(Frame::Ptr frame, const std::vector<cv::DMatch> &matches);

    void SetMap(Map::Ptr map) { map_ = map; }


    bool Track(Frame::Ptr new_frame);
    bool AddFrame(Frame::Ptr frame);
    bool StereoInit();
    enum class FrontendState { INITIALIZING, TRACKING_GOOD, LOST };
    FrontendState state_ = FrontendState::INITIALIZING;


private:
    Camera::Ptr camera_left_ = nullptr;
    Camera::Ptr camera_right_ = nullptr;

    int num_features_;
    double roi_top_fraction_; // چه کسری از بالای تصویر را حذف کنیم
    cv::Ptr<cv::ORB> orb_;
    cv::Ptr<cv::GFTTDetector> gftt_; // optional, اگر خواستیم
    // helper
    cv::Rect ComputeROI(const cv::Mat &img) const;
    void KeepBestKeypoints(std::vector<cv::KeyPoint> &kps, cv::Mat &descriptors, int n);

    Map::Ptr map_;
    long next_landmark_id_ = 0;

    Frame::Ptr current_frame_ = nullptr;
    Frame::Ptr last_frame_ = nullptr;
    cv::Mat K_;   // Optinal
    double baseline_;

};

} // namespace myslam
````

````python
%%writefile frontend.cpp

// src/frontend.cpp
#include "myslam/frontend.h"
#include <numeric>
#include <fstream>

namespace myslam {

Frontend::Frontend(int num_features, double roi_top_fraction)
    : num_features_(num_features), roi_top_fraction_(roi_top_fraction) {
    orb_ = cv::ORB::create(num_features_);
    gftt_ = cv::GFTTDetector::create(num_features_ * 2, 0.01, 20);
    K_ = (cv::Mat_<double>(3,3) <<
        707.0912, 0, 601.8873,
        0, 707.0912, 183.1104,
        0, 0, 1);
    baseline_ = 379.8145 / 707.0912;
}



cv::Rect Frontend::ComputeROI(const cv::Mat &img) const {
    if (roi_top_fraction_ <= 0.0) return cv::Rect(0,0, img.cols, img.rows);
    int y = static_cast<int>(img.rows * roi_top_fraction_);
    int h = img.rows - y;
    return cv::Rect(0, y, img.cols, h);
}

bool Frontend::StereoInit() {
    // اولین فریم را صرفاً به عنوان keyframe ذخیره کن
    current_frame_->SetPose(Sophus::SE3d());
    // map_->AddFrame(current_frame_);
    last_frame_ = current_frame_;
    return true;
}


void Frontend::KeepBestKeypoints(std::vector<cv::KeyPoint> &kps, cv::Mat &descriptors, int n) {
    if ((int)kps.size() <= n) return;
    // create indices sorted by response descending
    std::vector<int> idx(kps.size());
    std::iota(idx.begin(), idx.end(), 0);
    std::sort(idx.begin(), idx.end(), [&](int a, int b) {
        return kps[a].response > kps[b].response;
    });
    // pick top-n
    std::vector<cv::KeyPoint> kps2;
    cv::Mat desc2;
    kps2.reserve(n);
    for (int i = 0; i < n; ++i) {
        kps2.push_back(kps[idx[i]]);
        if (!descriptors.empty()) desc2.push_back(descriptors.row(idx[i]));
    }
    kps.swap(kps2);
    if (!desc2.empty()) descriptors = desc2;
}

bool Frontend::DetectAndCompute(Frame::Ptr frame) {
    if (!frame) return false;
    if (frame->left_img.empty()) return false;

    // ROI (crop top if configured)
    cv::Rect roi = ComputeROI(frame->left_img);
    cv::Mat left_roi = frame->left_img(roi);

    // Optional: create mask (exclude top small stripe or other rules)
    cv::Mat mask = cv::Mat::ones(left_roi.size(), CV_8U) * 255;
    // Example: exclude top 5% of roi (if you want more advanced mask, change here)
    // int exclude_h = static_cast<int>(0.05 * left_roi.rows);
    // if (exclude_h > 0) mask(cv::Rect(0,0,left_roi.cols, exclude_h)) = 0;

    // Detect+Compute on ROI
    std::vector<cv::KeyPoint> kps_left;
    cv::Mat desc_left;
    orb_->detectAndCompute(left_roi, mask, kps_left, desc_left);

    // adjust keypoint coordinates to original image coords
    for (auto &kp : kps_left) {
        kp.pt.x += roi.x;
        kp.pt.y += roi.y;
    }

    // Right image: detect on same ROI (or full image); here use same ROI for symmetry
    cv::Mat right_roi;
    std::vector<cv::KeyPoint> kps_right;
    cv::Mat desc_right;
    if (!frame->right_img.empty()) {
        right_roi = frame->right_img(roi);
        orb_->detectAndCompute(right_roi, mask, kps_right, desc_right);
        for (auto &kp : kps_right) {
            kp.pt.x += roi.x;
            kp.pt.y += roi.y;
        }
    }

    // Keep best by response
    KeepBestKeypoints(kps_left, desc_left, num_features_);
    if (!kps_right.empty()) KeepBestKeypoints(kps_right, desc_right, num_features_);

    // assign to frame
    frame->keypoints_left = std::move(kps_left);
    frame->descriptors_left = std::move(desc_left);
    frame->keypoints_right = std::move(kps_right);
    frame->descriptors_right = std::move(desc_right);

    return true;
}

bool Frontend::AddFrame(Frame::Ptr frame) {
    current_frame_ = frame;

    if (state_ == FrontendState::INITIALIZING) {
        bool success = StereoInit();
        if (success) {
            state_ = FrontendState::TRACKING_GOOD;
        }
        return success;
    }

    bool success = Track(frame);

    if (!success) {
        state_ = FrontendState::LOST;
        return false;
    }

    return true;
}

void Frontend::ComputeStereoDepth(Frame::Ptr frame, const std::vector<cv::DMatch> &matches) {

    const double fx = camera_left_->fx();
    const double fy = camera_left_->fy();
    const double cx = camera_left_->cx();
    const double cy = camera_left_->cy();
    const double baseline = camera_right_->baseline();

    frame->points_3d.clear();
    frame->points_3d.reserve(matches.size());

    for(const auto &m : matches) {
        const cv::KeyPoint &kpL = frame->keypoints_left[m.queryIdx];
        const cv::KeyPoint &kpR = frame->keypoints_right[m.trainIdx];

        double disparity = kpL.pt.x - kpR.pt.x;
        if(disparity <= 0.5) continue;

        double Z = fx * baseline / disparity;
        double X = (kpL.pt.x - cx) * Z / fx;
        double Y = (kpL.pt.y - cy) * Z / fy;

        frame->points_3d.emplace_back(Eigen::Vector3d(X, Y, Z));
    }

    std::cout << "[Depth] Valid 3D points: " << frame->points_3d.size() << std::endl;
}

void Frontend::SetCameras(Camera::Ptr cam_left, Camera::Ptr cam_right) {
    camera_left_ = cam_left;
    camera_right_ = cam_right;
}

void Frontend::MatchStereo(Frame::Ptr frame, std::vector<cv::DMatch> &matches) {
    matches.clear();
    if (!frame) return;
    if (frame->descriptors_left.empty() || frame->descriptors_right.empty()) return;

    // Step 1: brute-force matching
    cv::BFMatcher matcher(cv::NORM_HAMMING, true); // crossCheck = true
    std::vector<cv::DMatch> raw_matches;
    matcher.match(frame->descriptors_left, frame->descriptors_right, raw_matches);

    // Step 2: filtering based on epipolar constraint and angle similarity
    const double max_y_diff = 2.0;
    const double max_angle_diff = 30.0;

    for (auto& m : raw_matches) {
        const auto& kpL = frame->keypoints_left[m.queryIdx];
        const auto& kpR = frame->keypoints_right[m.trainIdx];

        if (std::fabs(kpL.pt.y - kpR.pt.y) > max_y_diff)
            continue;

        if (std::fabs(kpL.angle - kpR.angle) > max_angle_diff)
            continue;

        matches.push_back(m);
    }

    // Step 3: sort by descriptor distance
    std::sort(matches.begin(), matches.end(), [](const cv::DMatch &a, const cv::DMatch &b){
        return a.distance < b.distance;
    });

    // Step 4: keep only N best if needed
    int keep = std::min<int>(matches.size(), num_features_);
    matches.resize(keep);

    std::cout << "[Stereo Match] Raw: " << raw_matches.size()
              << "  After Filter: " << matches.size() << std::endl;

   // Step 5
    ComputeStereoDepth(frame, matches);

    // Step 6 === Create MapPoints and insert into Map ===
    for (size_t i = 0; i < frame->points_3d.size(); ++i) {

        static unsigned long next_landmark_id_ = 0;
        Eigen::Vector3d Pw = frame->points_3d[i];

        // Depth sanity check
        if (Pw[2] < 0.3 || Pw[2] > 80.0) continue;

        auto mp = std::make_shared<MapPoint>(next_landmark_id_++, Pw);
        mp->IncreaseObservation();
        // local association: برای محاسبه‌ی جزئیاتی مثل تعداد مشاهده‌ها،
        // کدام فریم‌ها آن را دیده‌اند، نمایش‌های محلی (draw) و تولید فایل‌های debug) نیاز داری بدانیم کدام فریم‌ها آن MP را دیده‌اند
        // فقط برای track کردن
        frame->AddLandmark(mp);
        // global storage:
        // شامل تمام نقاطی که ارزش نقشه شدن دارند
        // ورودی برای backend و BA
        map_->InsertMapPoint(mp);
    }

    // Mark this frame as keyframe (simple version)
    map_->InsertKeyFrame(frame);

    std::cout << "[MAP] KeyFrames: " << map_->KeyframeCount()
              << "  Landmarks: " << map_->MapPointsCount() << std::endl;
}

bool Frontend::Track(Frame::Ptr frame)
{
    if (last_frame_ == nullptr) {
        frame->SetPose(Sophus::SE3d());
        last_frame_ = frame;
        return true;
    }

    // 1. Match last_frame_ with current frame
    std::vector<cv::DMatch> matches;
    cv::BFMatcher matcher(cv::NORM_HAMMING, true);
    matcher.match(last_frame_->descriptors_left, frame->descriptors_left, matches);

    if (matches.size() < 10) {
        std::cout << "[Track] Not enough matches!\n";
        return false;
    }
    // ////// VO:

    // pts1 and pts2 are two sets of points taken from the left image
    // of the previous frame and the left image of the current frame.
    //
    // This means:
    // - corresponding pixels between the two frames have been found
    // - there is an underlying 3D camera motion behind these pixel correspondences

    // 2. Build points
    std::vector<cv::Point2f> pts1, pts2;
    for (auto &m : matches) {
        pts1.push_back(last_frame_->keypoints_left[m.queryIdx].pt);
        pts2.push_back(frame->keypoints_left[m.trainIdx].pt);
    }
    // Each point correspondence that remains after RANSAC
    // is considered a tracked point.

    // Matches = Tracked Points Within Frames

    // 3. Essential matrix
    // The Essential Matrix describes the geometric relationship between:
    // - two consecutive images
    // - captured by a calibrated camera
    // - undergoing rigid motion (R, t)


    cv::Mat E = cv::findEssentialMat(pts1, pts2, K_, cv::RANSAC);

    // 4. Recover pose
    // This function converts the Essential Matrix into an estimate of R and t.
    //
    // It uses the epipolar geometry of the matched 2D points
    // to recover the relative camera motion.
    //
    // This is NOT PnP, because PnP requires known 3D points.
    // At this stage, we are still working with 2D-2D correspondences.


    cv::Mat R, t;
    cv::recoverPose(E, pts1, pts2, K_, R, t);

    Eigen::Matrix3d R_eigen;
    Eigen::Vector3d t_eigen;

    cv::Mat_<double> RR = R;
    cv::Mat_<double> tt = t;

    for(int i=0;i<3;i++)
        for(int j=0;j<3;j++)
            R_eigen(i,j) = RR(i,j);

    for(int i=0;i<3;i++)
        t_eigen(i) = tt(i);


    // ----- Add here -----

    // Create SE3 pose for this frame
    Eigen::Matrix4d T_curr = Eigen::Matrix4d::Identity();
    T_curr.block<3,3>(0,0) = R_eigen;
    T_curr.block<3,1>(0,3) = t_eigen;

    // Euler angles
    double yaw1, pitch1, roll1;
    roll1 = atan2(R_eigen(2,1), R_eigen(2,2));
    pitch1 = -asin(R_eigen(2,0));
    yaw1 = atan2(R_eigen(1,0), R_eigen(0,0));

    // Translation (relative pose)
    double tx = t_eigen(0);
    double ty = t_eigen(1);
    double tz = t_eigen(2);

    // Print
    std::cout << "[Pose] yaw1: " << yaw1
              << " | pitch1: " << pitch1
              << " | roll1: " << roll1 << std::endl;

    std::cout << "[Pose] translation: "
              << tx << ", " << ty << ", " << tz << std::endl;


    // -------------------------------------------
    // ACCUMULATE GLOBAL TRAJECTORY
    // -------------------------------------------
    static Eigen::Matrix4d T_global = Eigen::Matrix4d::Identity();
    T_global = T_global * T_curr;

    // Extract global R,t
    Eigen::Matrix3d Rg = T_global.block<3,3>(0,0);
    Eigen::Vector3d tg = T_global.block<3,1>(0,3);

    // -------------------------------------------
    // Convert Rotation → Yaw Pitch Roll
    // -------------------------------------------
    double yaw   = atan2(Rg(1,0), Rg(0,0));
    double pitch = atan2(-Rg(2,0), sqrt(Rg(2,1)*Rg(2,1) + Rg(2,2)*Rg(2,2)));
    double roll  = atan2(Rg(2,1), Rg(2,2));

    // -------------------------------------------
    // PRINT POSE INFORMATION
    // -------------------------------------------
    //std::cout << "\n[POSE] Frame: " << frame_id << "\n";
    std::cout << "Translation (global): " << tg.transpose() << "\n";
    std::cout << "Yaw: " << yaw << " | Pitch: " << pitch << " | Roll: " << roll << "\n\n";

    // ............................................

    // 5. Convert to Sophus SE3
    // ساخت Sophus
    Sophus::SE3d relative(R_eigen, t_eigen);


    // Pose = last_pose * relative
    Sophus::SE3d last_pose = last_frame_->Pose();
    // frame->SetPose(last_pose * relative);
    //frame->SetPose(last_frame_->Pose() * relative);


    Sophus::SE3d T_global_se3( T_global.block<3,3>(0,0), T_global.block<3,1>(0,3) );
    frame->SetPose(T_global_se3);

    last_frame_ = frame;

    return true;
}


} // namespace myslam
````

**So, What Is This Method?**

**Essential Matrix Pose Estimation**
or  
**5-Point Algorithm for Motion Estimation**

This method:

- Works only with corresponding points from two frames, without requiring known 3D points.
- Estimates the relative camera pose based on **epipolar geometry**.

---

**Why Is It Not PnP?**

PnP requires:

- 3D points: \((X, Y, Z)\)
- Corresponding image points: \((u, v)\)
- A **3D-to-2D projection** relationship

But here:

- No 3D MapPoints are used.
- Only corresponding points from two frames are used.

Therefore, this is **not PnP**.

---

**Why Is It Not Triangulation?**

Triangulation means:

- having two observations of the same point,
- knowing the relative camera motion \((R,t)\),
- and reconstructing the corresponding **3D point**.

In this stage:

- No 3D point is actually reconstructed.
- Only the relative camera motion \((R,t)\) is estimated.

Therefore, this is **not Triangulation** either.

````python
import open3d as o3d
import numpy as np

# بارگذاری point cloud و trajectory
pcd = o3d.io.read_point_cloud("/content/MySLAM_Ch13/build/map_1770714878.ply")
traj = np.loadtxt("/content/MySLAM_Ch13/build/trajectory.txt")[:, :3]

# --- تطبیق مسیر با map ---
pcd_points = np.asarray(pcd.points)
pcd_min = pcd_points.min(axis=0)
pcd_max = pcd_points.max(axis=0)
pcd_center = (pcd_max + pcd_min) / 2
pcd_range = pcd_max - pcd_min

traj_center = traj.mean(axis=0)
traj_range = traj.max(axis=0) - traj.min(axis=0)

# scale factor برای هر محور
scale = pcd_range / traj_range
scale_factor = scale.mean()  # می‌تونیم میانگین محورها رو بگیریم

# اعمال scale و translation
traj_aligned = (traj - traj_center) * scale_factor + pcd_center

# --- ساخت LineSet برای نمایش مسیر ---
lines = [[i, i + 1] for i in range(len(traj_aligned) - 1)]
line_set = o3d.geometry.LineSet(
    points=o3d.utility.Vector3dVector(traj_aligned),
    lines=o3d.utility.Vector2iVector(lines)
)

# نمایش با Plotly در Colab/Jupyter
o3d.visualization.draw_plotly([pcd, line_set])
````
